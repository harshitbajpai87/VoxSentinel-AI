## Protocol
VoxCeleb1 verification trials; threshold swept for EER; include TTS/cloned impostors to show similarity is not spoof-proof.

```python
import torch, torchaudio
def embed(p):
    w, sr = torchaudio.load(p)
    if sr != 16000: w = torchaudio.functional.resample(w, sr, 16000)
    return enc.encode_batch(w).squeeze().detach().cpu()

def cosine(a, b): return float(torch.nn.functional.cosine_similarity(a, b, dim=0))
# pairs = [(path1, path2, same?) ...] built from VoxCeleb trials
print("score pairs -> sweep threshold -> report EER and minDCF")
```
